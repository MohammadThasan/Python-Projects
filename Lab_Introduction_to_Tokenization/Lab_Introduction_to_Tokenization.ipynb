# Lab: Introduction to Tokenization

## Introduction
Tokenization is the process of splitting text into smaller units called tokens. These tokens can be words, subwords, characters, or punctuation marks. Tokenization is a crucial first step in many Natural Language Processing (NLP) tasks, such as text classification, machine translation, and language modeling.

In this lab, you will learn about several common tokenization methods, understand their advantages and disadvantages, and see simple Python examples for each.


## 1. What Is Tokenization?

Tokenization transforms a raw text string into a list of tokens. Depending on the method, tokens might be:

- **Words**: e.g., "reading"
- **Punctuation**: e.g., "!"
- **Characters**: e.g., "r", "e", "a", ...
- **Subwords**: e.g., "read", "ing"

Choosing the right tokenization strategy affects the size of the vocabulary, handling of out-of-vocabulary (OOV) words, and model performance.


## 2. Whitespace Tokenization

**Concept**: Split text on whitespace characters (spaces, tabs, newlines).

- **Pros**: Very fast and easy to implement.
- **Cons**: Punctuation remains attached to words (e.g., "hello!" stays as one token).


```python
# Sample text
txt = "Hello, world!\nWelcome to tokenization."

# Whitespace tokenization
tokens = txt.split()
print(tokens)
# Output: ['Hello,', 'world!', 'Welcome', 'to', 'tokenization.']
```

## 3. Word-Level Tokenization

**Concept**: Use simple rules or regular expressions to separate words and punctuation.

- **Pros**: More precise separation of punctuation.
- **Cons**: Can lead to a large vocabulary and OOV words.


```python
import re

# Sample text
txt = "Don't stop believing!"

# Regex to capture words and punctuation
tokens = re.findall(r"[A-Za-z]+(?:'[A-Za-z]+)?|[.,!?]", txt)
print(tokens)
# Output: ["Don't", 'stop', 'believing', '!']
```

## 4. Character-Level Tokenization

**Concept**: Treat every character (including punctuation and spaces) as a token.

- **Pros**: No OOV issues; small vocabulary (all possible characters).
- **Cons**: Very long token sequences; harder for models to learn context.


```python
# Sample text
txt = "Hi!"

# Character-level tokens
tokens = list(txt)
print(tokens)
# Output: ['H', 'i', '!']
```

## 5. Subword Tokenization

Subword methods break words into smaller, meaningful pieces. They balance vocabulary size and the ability to handle rare words.

### 5.1 Byte Pair Encoding (BPE)

**Concept**: Iteratively merge the most frequent pairs of characters or subwords.


```python
# Install the library (if needed)
# pip install tokenizers
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import Whitespace

# Initialize tokenizer
tokenizer = Tokenizer(BPE(unk_token="[UNK]"))
# Use whitespace to split before BPE
tokenizer.pre_tokenizer = Whitespace()

# Prepare trainer
trainer = BpeTrainer(vocab_size=30, special_tokens=["[UNK]", "[CLS]", "[SEP]"])

# Train on example sentences
sentences = ["I love NLP.", "Tokenization is fun!"]
tokenizer.train_from_iterator(sentences, trainer)

# Encode text
output = tokenizer.encode("Tokenization is powerful.")
print(output.tokens)
# Example output: ['Token', 'ization', 'is', 'powerful', '.']
```

## 6. Summary

| Method                   | Vocabulary Size    | Handles OOV       | Sequence Length |
|--------------------------|--------------------|-------------------|-----------------|
| Whitespace               | Small              | No                | Short           |
| Word-Level               | Large              | Some OOV          | Medium          |
| Character-Level          | Very Small         | Yes               | Long            |
| Subword (e.g., BPE)      | Medium             | Yes               | Medium          |

Each tokenization method has trade-offs. Choose based on your task requirements and computational resources.
